```python
from pathlib import Path
import pandas as pd

data_dir = Path("../data/1mil")

movies = pd.read_csv(
    data_dir / "movies.dat",
    sep="::",
    engine="python",
    encoding="latin-1",
    names=["movie_id", "title", "genres"]
)

ratings = pd.read_csv(
    data_dir / "ratings.dat",
    sep="::",
    engine="python",
    encoding="latin-1",
    names=["user_id", "movie_id", "rating", "timestamp"]
)

users = pd.read_csv(
    data_dir / "users.dat",
    sep="::",
    engine="python",
    encoding="latin-1",
    names=["user_id", "gender", "age", "occupation", "zip_code"]
)

ratings["datetime"] = pd.to_datetime(ratings["timestamp"], unit="s")

age_map = {
    1: "Under 18",
    18: "18-24",
    25: "25-34",
    35: "35-44",
    45: "45-49",
    50: "50-55",
    56: "56+"
}

occupation_map = {
    0: "other/not specified",
    1: "academic/educator",
    2: "artist",
    3: "clerical/admin",
    4: "college/grad student",
    5: "customer service",
    6: "doctor/health care",
    7: "executive/managerial",
    8: "farmer",
    9: "homemaker",
    10: "K-12 student",
    11: "lawyer",
    12: "programmer",
    13: "retired",
    14: "sales/marketing",
    15: "scientist",
    16: "self-employed",
    17: "technician/engineer",
    18: "tradesman/craftsman",
    19: "unemployed",
    20: "writer"
}

users["age_group"] = users["age"].map(age_map)
users["occupation_name"] = users["occupation"].map(occupation_map)

print("Movies:", movies.shape)
print("Ratings:", ratings.shape)
print("Users:", users.shape)

display(movies.head(10))
display(ratings.head(10))
display(users.head(10))

print("Rating distribution:")
display(ratings["rating"].value_counts().sort_index())

print("Age distribution:")
display(users["age_group"].value_counts().sort_index())

print("Gender distribution:")
display(users["gender"].value_counts())

print("Occupation distribution:")
display(users["occupation_name"].value_counts())

ratings_with_movies = ratings.merge(movies, on="movie_id", how="left")
full_data = ratings_with_movies.merge(users, on="user_id", how="left")

print("Joined dataset:")
display(full_data.head(10))

movie_stats = (
    ratings
    .groupby("movie_id")
    .agg(
        num_ratings=("rating", "count"),
        avg_rating=("rating", "mean")
    )
    .merge(movies, on="movie_id", how="left")
    .sort_values("num_ratings", ascending=False)
)

print("Top 10 most-rated movies:")
display(movie_stats.head(10))

print("Top 10 highest-rated movies with at least 100 ratings:")
display(
    movie_stats
    .query("num_ratings >= 100")
    .sort_values("avg_rating", ascending=False)
    .head(10)
)

genres_exploded = (
    movies
    .assign(genre=movies["genres"].str.split("|"))
    .explode("genre")
)

print("Genre counts:")
display(genres_exploded["genre"].value_counts())

genre_rating_stats = (
    ratings_with_movies
    .assign(genre=ratings_with_movies["genres"].str.split("|"))
    .explode("genre")
    .groupby("genre")
    .agg(
        num_ratings=("rating", "count"),
        avg_rating=("rating", "mean")
    )
    .sort_values("num_ratings", ascending=False)
)

print("Genre rating stats:")
display(genre_rating_stats)

ratings["rating"].value_counts().sort_index().plot(
    kind="bar",
    title="Rating distribution",
    xlabel="Rating",
    ylabel="Count"
)
```

```python
from pathlib import Path
import pandas as pd

# ---------- mappings from README ----------

age_map = {
    1: "Under 18",
    18: "18-24",
    25: "25-34",
    35: "35-44",
    45: "45-49",
    50: "50-55",
    56: "56+"
}

occupation_map = {
    0: "other/not specified",
    1: "academic/educator",
    2: "artist",
    3: "clerical/admin",
    4: "college/grad student",
    5: "customer service",
    6: "doctor/health care",
    7: "executive/managerial",
    8: "farmer",
    9: "homemaker",
    10: "K-12 student",
    11: "lawyer",
    12: "programmer",
    13: "retired",
    14: "sales/marketing",
    15: "scientist",
    16: "self-employed",
    17: "technician/engineer",
    18: "tradesman/craftsman",
    19: "unemployed",
    20: "writer"
}

users = users.copy()
users["age_group"] = users["age"].map(age_map)
users["occupation_name"] = users["occupation"].map(occupation_map)

# ---------- clean ratings ----------

ratings_clean = ratings.copy()
ratings_clean["datetime"] = pd.to_datetime(ratings_clean["timestamp"], unit="s")

print("Before filtering")
print("ratings:", len(ratings_clean))
print("users:", ratings_clean["user_id"].nunique())
print("movies:", ratings_clean["movie_id"].nunique())

min_user_ratings = 20
min_movie_ratings = 20

old_size = None

while old_size != len(ratings_clean):
    old_size = len(ratings_clean)

    user_counts = ratings_clean["user_id"].value_counts()
    users_to_keep = user_counts[user_counts >= min_user_ratings].index
    ratings_clean = ratings_clean[ratings_clean["user_id"].isin(users_to_keep)]

    movie_counts = ratings_clean["movie_id"].value_counts()
    movies_to_keep = movie_counts[movie_counts >= min_movie_ratings].index
    ratings_clean = ratings_clean[ratings_clean["movie_id"].isin(movies_to_keep)]

ratings_clean = ratings_clean.reset_index(drop=True)

print("\nAfter sparse filtering")
print("ratings:", len(ratings_clean))
print("users:", ratings_clean["user_id"].nunique())
print("movies:", ratings_clean["movie_id"].nunique())

# ---------- keep all ratings, but add interpretation labels ----------

ratings_clean["preference_label"] = ratings_clean["rating"].map({
    1: "strong_dislike",
    2: "dislike",
    3: "neutral",
    4: "like",
    5: "strong_like"
})

# Useful later if you want a binary label,
# but rating 3 is left as unknown instead of being forced into 0 or 1.
ratings_clean["liked_binary"] = ratings_clean["rating"].map({
    1: 0,
    2: 0,
    3: None,
    4: 1,
    5: 1
})

print("\nRating distribution after filtering")
print(ratings_clean["rating"].value_counts().sort_index())

print("\nPreference labels")
print(ratings_clean["preference_label"].value_counts())

# ---------- fully merged clean dataset ----------

ratings_clean_full = (
    ratings_clean
    .merge(movies, on="movie_id", how="left")
    .merge(users, on="user_id", how="left")
)

ratings_clean_full = ratings_clean_full[
    [
        "user_id",
        "movie_id",
        "rating",
        "preference_label",
        "liked_binary",
        "timestamp",
        "datetime",
        "title",
        "genres",
        "gender",
        "age",
        "age_group",
        "occupation",
        "occupation_name",
        "zip_code"
    ]
]

print("\nMerged clean dataset")
print("rows:", len(ratings_clean_full))
print("users:", ratings_clean_full["user_id"].nunique())
print("movies:", ratings_clean_full["movie_id"].nunique())

display(ratings_clean_full.head())

# ---------- save one fully merged NDJSON file ----------

output_dir = Path("processed")
output_dir.mkdir(parents=True, exist_ok=True)

output_file = output_dir / "ratings_clean_full.ndjson"

ratings_clean_full.to_json(
    output_file,
    orient="records",
    lines=True,
    force_ascii=False
)

print("\nSaved file:")
print(output_file.resolve())
```

```python
from pathlib import Path
import pandas as pd
import random

# ---------- load clean merged dataset ----------

input_file = Path("processed/ratings_clean_full.ndjson")

ratings_clean_full = pd.read_json(
    input_file,
    orient="records",
    lines=True
)

# Turn "Drama|Comedy|Romance" into ["Drama", "Comedy", "Romance"]
ratings_clean_full["genres"] = ratings_clean_full["genres"].apply(
    lambda x: x.split("|") if isinstance(x, str) else []
)

print("Loaded clean dataset")
print("rows:", len(ratings_clean_full))
print("users:", ratings_clean_full["user_id"].nunique())
print("movies:", ratings_clean_full["movie_id"].nunique())

display(ratings_clean_full.head())


# ---------- pairwise construction settings ----------

min_rating_difference = 2
max_pairs_per_user = 100

random_seed = 42
random.seed(random_seed)


# ---------- create pairwise comparisons ----------

pair_rows = []

for user_id, user_df in ratings_clean_full.groupby("user_id"):
    user_df = user_df.sort_values("timestamp")

    user_pairs = []

    ratings_list = user_df[
        [
            "movie_id",
            "rating",
            "timestamp",
            "title",
            "genres"
        ]
    ].to_dict("records")

    for a in range(len(ratings_list)):
        for b in range(a + 1, len(ratings_list)):
            item_a = ratings_list[a]
            item_b = ratings_list[b]

            rating_diff = item_a["rating"] - item_b["rating"]

            # Drop ties and near-ties
            if abs(rating_diff) < min_rating_difference:
                continue

            if rating_diff > 0:
                preferred = item_a
                less_preferred = item_b
            else:
                preferred = item_b
                less_preferred = item_a

            user_pairs.append({
                "user_id": user_id,

                "preferred_movie_id": preferred["movie_id"],
                "less_preferred_movie_id": less_preferred["movie_id"],

                "preferred_rating": preferred["rating"],
                "less_preferred_rating": less_preferred["rating"],
                "rating_difference": abs(rating_diff),

                "preferred_title": preferred["title"],
                "less_preferred_title": less_preferred["title"],

                "preferred_genres": preferred["genres"],
                "less_preferred_genres": less_preferred["genres"]
            })

    if len(user_pairs) > max_pairs_per_user:
        user_pairs = random.sample(user_pairs, max_pairs_per_user)

    pair_rows.extend(user_pairs)


pairs = pd.DataFrame(pair_rows)

print("Pairwise dataset")
print("rows:", len(pairs))
print("users:", pairs["user_id"].nunique())

display(pairs.head(20))

print("\nRating difference distribution:")
display(pairs["rating_difference"].value_counts().sort_index())


# ---------- save pairwise dataset ----------

output_file = Path("processed/pairwise_preferences.ndjson")

pairs.to_json(
    output_file,
    orient="records",
    lines=True,
    force_ascii=False
)

print("\nSaved pairwise dataset:")
print(output_file.resolve())
```

```python
from pathlib import Path
import pandas as pd
import random

# ---------- load clean merged dataset ----------

input_file = Path("processed/ratings_clean_full.ndjson")

ratings_clean_full = pd.read_json(
    input_file,
    orient="records",
    lines=True
)

# Make sure genres are arrays
ratings_clean_full["genres"] = ratings_clean_full["genres"].apply(
    lambda x: x.split("|") if isinstance(x, str) else x
)

print("Loaded clean dataset")
print("rows:", len(ratings_clean_full))
print("users:", ratings_clean_full["user_id"].nunique())
print("movies:", ratings_clean_full["movie_id"].nunique())

display(ratings_clean_full.head())


# ---------- pairwise construction choices ----------

min_rating_difference = 2
max_pairs_per_user = 100
random_seed = 42

random.seed(random_seed)


# ---------- create pairwise comparisons ----------

pair_rows = []

for user_id, user_df in ratings_clean_full.groupby("user_id"):
    user_df = user_df.sort_values("timestamp")

    user_pairs = []

    ratings_list = user_df[
        [
            "movie_id",
            "rating",
            "timestamp",
            "title",
            "genres"
        ]
    ].to_dict("records")

    for a in range(len(ratings_list)):
        for b in range(a + 1, len(ratings_list)):
            item_a = ratings_list[a]
            item_b = ratings_list[b]

            rating_diff = item_a["rating"] - item_b["rating"]

            # 0 = tie, 1 = near-tie.
            # For Step 1, keep only clear preferences.
            if abs(rating_diff) < min_rating_difference:
                continue

            if rating_diff > 0:
                preferred = item_a
                less_preferred = item_b
            else:
                preferred = item_b
                less_preferred = item_a

            user_pairs.append({
                "user_id": user_id,

                "preferred_movie_id": preferred["movie_id"],
                "less_preferred_movie_id": less_preferred["movie_id"],

                "preferred_rating": preferred["rating"],
                "less_preferred_rating": less_preferred["rating"],
                "rating_difference": abs(rating_diff),

                "preferred_title": preferred["title"],
                "less_preferred_title": less_preferred["title"],

                "preferred_genres": preferred["genres"],
                "less_preferred_genres": less_preferred["genres"]
            })

    # Avoid pair explosion and avoid very active users dominating
    if len(user_pairs) > max_pairs_per_user:
        user_pairs = random.sample(user_pairs, max_pairs_per_user)

    pair_rows.extend(user_pairs)


pairs = pd.DataFrame(pair_rows)

print("\nPairwise dataset")
print("rows:", len(pairs))
print("users:", pairs["user_id"].nunique())

display(pairs.head(20))

print("\nRating difference distribution:")
display(pairs["rating_difference"].value_counts().sort_index())

print("\nPairs per user:")
pairs_per_user = pairs.groupby("user_id").size()
display(pairs_per_user.describe())


# ---------- save pairwise dataset ----------

output_file = Path("processed/pairwise_preferences.ndjson")

pairs.to_json(
    output_file,
    orient="records",
    lines=True,
    force_ascii=False
)

print("\nSaved pairwise dataset:")
print(output_file.resolve())
```
